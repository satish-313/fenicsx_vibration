```python
gmsh.initialize()
gmsh.model.add("test1")
gmsh.model.occ.importShapes("./cad/test_cube.step")
gmsh.model.occ.synchronize()
```

```python
volumes = gmsh.model.occ.getEntities(dim=3)
print(len(volumes))

surfaces = gmsh.model.occ.getEntities(dim=2)
print(len(surfaces))
```

```python
gmsh.model.occ.fragment(volumes, [])
gmsh.model.occ.synchronize()

volumes = gmsh.model.occ.getEntities(dim=3)
if len(volumes) == 2:
    print("Ok")
    
surfaces = gmsh.model.occ.getEntities(dim=2)
print(len(surfaces))
```

```python
def get_volume_surfaces(volume_tag):
    boundary = gmsh.model.getBoundary(
        [(3, volume_tag)],
        oriented=False,
        recursive=False
    )

    return {tag for dim, tag in boundary if dim == 2}


def shared_surfaces(volume1, volume2):
    surfaces1 = get_volume_surfaces(volume1)
    surfaces2 = get_volume_surfaces(volume2)

    return surfaces1.intersection(surfaces2)

shared = shared_surfaces(1, 2)

print("Shared surfaces:", shared)
```
