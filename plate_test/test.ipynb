```python
import gmsh
from mpi4py import MPI
from dolfinx.io import gmsh as gmshio

gmsh.initialize()
gmsh.model.add("plate")
gmsh.model.occ.importShapes("circular_plate-Body.step")
gmsh.model.occ.synchronize()

gmsh.option.setNumber("Mesh.CharacteristicLengthMin", 0.1)
gmsh.option.setNumber("Mesh.CharacteristicLengthMax", 0.1)

# Get all 3D volumes
volumes = gmsh.model.getEntities(dim=3)

volume_tags = [v[1] for v in volumes]
# Create physical group for the solid domain (tag=1)
gmsh.model.addPhysicalGroup(3, volume_tags, tag=1, name="plate_volume")

# Optional: Add physical group for all outer boundaries (dim=2)
surfaces = gmsh.model.getEntities(dim=2)
surface_tags = [s[1] for s in surfaces]
gmsh.model.addPhysicalGroup(2, surface_tags, tag=2, name="plate_surfaces")

gmsh.model.mesh.generate(3)

mesh, cell_tags, facet_tags = gmshio.model_to_mesh(
    gmsh.model, MPI.COMM_WORLD, rank=0,gdim=3
)

gmsh.finalize()
```
